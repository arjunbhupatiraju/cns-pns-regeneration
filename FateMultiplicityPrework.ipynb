```python
import sys, inspect, json, glob
V1 = '/content/drive/MyDrive/CNS_PNS_Trajectory_Stability'
sys.path.insert(0, V1 + '/src')

import fatestability.inference as inf
print("--- inference exports ---")
print([x for x in dir(inf) if not x.startswith('_')])

print("\n--- signatures ---")
for name in dir(inf):
    if name.startswith('_'):
        continue
    obj = getattr(inf, name)
    if callable(obj):
        try:
            print(f"{name}{inspect.signature(obj)}")
        except (ValueError, TypeError):
            pass

print("\n--- contracts ---")
for p in glob.glob(V1 + '/**/*contract*.json', recursive=True):
    print(p)
    print(list(json.load(open(p)).keys()))
```
